# Iris DataSet With Multiple Models and Inferencing 

```python
import pandas as pd
import numpy as np
import mlflow
import mlflow.sklearn   
import mlflow.xgboost
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from mlflow.sklearn import log_model
from sklearn.datasets import make_classification    
from xgboost.sklearn import XGBClassifier
from imblearn.combine import SMOTETomek
```

```python
X,y = load_iris(return_X_y=True)

```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,stratify=y,random_state=42)

```

```python
X,y = load_iris(return_X_y=True)

lr_params= {
    'penalty':'l2',
    'random_state':88,
    'solver':'newton-cg',
    'max_iter':1000,
}

lr = LogisticRegression(**lr_params)

lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
acc = accuracy_score(y_test,y_pred)
cross_val_score(lr,X,y,cv=5)
```

```python
y.shape,y_train.shape
```

```python
model_info =[
    ("Logistic Regression",lr_params,LogisticRegression()),
     ("Random Forest",{"n_estimators":1000,"random_state":88},RandomForestClassifier())
]
```

```python
for i in model_info:
    print(i)
```

```python
reports = []

for model_name, params, model  in model_info:
  
    model.set_params(**params)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    report = classification_report(y_test, y_pred, output_dict=True)
    reports.append(report)
```

```python
reports[0][2]
```

```python
mlflow.set_experiment("MLflow Starter - Tutorial 2")

mlflow.set_tracking_uri("http://localhost:5000")

for i, element in enumerate(model_info):
    model_name = element[0]
    params = element[1]
    model = element[2]
    report = reports[i]
    
    with mlflow.start_run(run_name=f"Tutorial 2 - {model_name} Training") as run:

        mlflow.log_params(params)
        print("0",report[2].values())
        mlflow.log_metrics({
            'accuracy': report[1],
            'recall_class_2': report[2]['2']['recall'],
            'recall_class_1': report[2]['1']['recall'],
            'recall_class_0': report[2]['0']['recall'],
            'f1_score_macro': report[2]['macro avg']['f1-score']
        }   )
        signature = mlflow.models.infer_signature(X_train, model.predict(X_train))
        mlflow.sklearn.log_model(model, artifact_path=f"{model_name}_model", signature=signature,input_example=X_train)
        
        
```
